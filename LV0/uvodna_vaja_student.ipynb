### HISTOGRAM

```python
import matplotlib.pyplot as plt
import numpy as np

ocene = [8, 7, 9, 6, 10, 8, 7, 9, 8, 6, 7, 8, 9, 7, 8, 6, 10, 9, 8, 7]

plt.figure(figsize=(10, 6))

# meje: 5.5–6.5, 6.5–7.5, ... → stolpci so centrirani na 6, 7, 8, 9, 10
bins = np.arange(5.5, 11.5, 1)

plt.hist(ocene, bins=bins, edgecolor='black', alpha=0.7)

plt.title('Porazdelitev ocen študentov')
plt.xlabel('Ocena')
plt.ylabel('Število študentov')
plt.grid(axis='y', alpha=0.3)

plt.xticks(range(6, 11))  # 6, 7, 8, 9, 10

plt.savefig('Rezultati/graf_histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

### KAZALČNI DIAGRAM

```python
import numpy as np
primer_amp = 1.0
primer_phi_deg = 35
primer_phi_rad = np.deg2rad(primer_phi_deg)

primer_x = primer_amp * np.cos(primer_phi_rad)
primer_y = primer_amp * np.sin(primer_phi_rad)

fig, ax = plt.subplots()
ax.quiver(0, 0, primer_x, primer_y,
          angles="xy", scale_units="xy", scale=1)
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 1.2)
ax.set_aspect("equal")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True)
plt.savefig('Rezultati/graf_KD.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Naloga
Naredite izris kazalčnega diagrama in ga shranite v mapo `Rezultati` pod ime `graf_KD_2.png`. Uporabite datoteki `Podatki/amp_u.csv` in `Podatki/fi_u.csv` in pri obeh vzemite vrstico, ki se skalda z zadnjima dvema števkama vaše vpisne številke. Pri izrisu uporabite funkcijo `quiver` in `amp_u` pomnožite s 100.

Primer: vpisna številka = 64220306 --> vrstica 6

```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import os

# Kreiramo folder Rezultati ako ne postoji
os.makedirs("Rezultati", exist_ok=True)

# Učitavanje fajlova
amp_df = pd.read_csv("Podatki/amp_u.csv", header=None)
fi_df = pd.read_csv("Podatki/fi_u.csv", header=None)

red_indeks = 78
if red_indeks >= len(amp_df):
    red_indeks = len(amp_df) - 1

# Sigurno pretvaranje u numeričke vrednosti (sprečava grešku sa stringovima)
amp_vals = pd.to_numeric(amp_df.iloc[red_indeks], errors='coerce').values * 100
fi_vals = pd.to_numeric(fi_df.iloc[red_indeks], errors='coerce').values

x = np.arange(len(amp_vals))
y = np.zeros_like(x)

u = amp_vals * np.cos(fi_vals)
v = amp_vals * np.sin(fi_vals)

# Crtanje i snimanje grafa
plt.figure(figsize=(8, 6))
plt.quiver(x, y, u, v, angles='xy', scale_units='xy', scale=1, color='b')
plt.title("Kazalčni diagram")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.savefig('Rezultati/graf_KD_2.png', dpi=300, bbox_inches='tight')
plt.close()

print("Graf je uspešno kreiran i sačuvan u Rezultati/graf_KD_2.png!")
```

### Naloga
Naredite izris poteka toka in njegova kota, in ga shranite v mapo `Rezultati` pod ime `graf_amp_fi_i.png`, na dveh ločenih grafih. Pri tem uporabite csv datoteki `amp_i` in `fi_i` ter si pomagajte s funkcijo `subplot`. Narišite za 20 ms (oziroma za 20 vrstic iz csv datoteke) in začnite izris pri vrstici, ki ju predstavljata zadnji dve števki vpisne številke (enako kot pri zgornji nalogi). Iz tega intervala najdite in zapišite najnižjo vrednost amplitude v spremenljivko `min_amp_i`.

Primer: vpisna številka = 64220306 --> od vrstice 6 do 26

```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import os

# Osiguravamo postojanje foldera Rezultati
os.makedirs("Rezultati", exist_ok=True)

# Učitavanje CSV fajlova
try:
    amp_df = pd.read_csv("Podatki/amp_i.csv", header=None)
    fi_df = pd.read_csv("Podatki/fi_i.csv", header=None)
except FileNotFoundError:
    amp_df = pd.read_csv("amp_i.csv", header=None)
    fi_df = pd.read_csv("fi_i.csv", header=None)

# Početni red prema zadnje dve cifre vpisne (78), uzimamo 20 redova (od 78 do 98)
start_red = 78
kraj_red = start_red + 20

if kraj_red > len(amp_df):
    kraj_red = len(amp_df)
    start_red = max(0, kraj_red - 20)

# Izdvajanje podataka za dati interval
amp_interval = pd.to_numeric(amp_df.iloc[start_red:kraj_red, 0], errors='coerce').values
fi_interval = pd.to_numeric(fi_df.iloc[start_red:kraj_red, 0], errors='coerce').values

# Nalazimo najnižu vrednost amplitude u ovom intervalu i čuvamo u min_amp_i
min_amp_i = float(np.min(amp_interval))

# Kreiranje subplots (2 odvojena grafa)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8))

# Prvi graf: Tok toka (amplituda)
ax1.plot(amp_interval, color='r', marker='o')
ax1.set_title("Potek toka - Amplituda")
ax1.set_ylabel("Amp")
ax1.grid(True)

# Drugi graf: Kota
ax2.plot(fi_interval, color='g', marker='x')
ax2.set_title("Kota toka")
ax2.set_xlabel("Uzorci (20 ms / 20 vrstic)")
ax2.set_ylabel("Kot")
ax2.grid(True)

plt.tight_layout()

# Čuvanje grafa pod traženim imenom u folder Rezultati
plt.savefig('Rezultati/graf_amp_fi_i.png', dpi=300, bbox_inches='tight')
plt.close()

print(f"Uspešno završeno! min_amp_i = {min_amp_i}")
```

```python
# ============================================================
# PREVERJANJE VAJE - TE CELICE NE SPREMINJAJTE
# ============================================================
# Ko zakljucite vajo, samo pozenite to celico.
# Celica poslje vase KONCNE REZULTATE na streznik za preverjanje.
# Pravilne resitve niso shranjene v tem notebooku in se vam ne prikazejo.
# Ce so obvezni rezultati pravilni, boste dobili kodo za eUcilnico.
# Grafi se ocenijo samo informativno in NE vplivajo na izdajo kode.
# ============================================================

import json
import base64
import os
import urllib.request
import urllib.error

# UCITELJ: pred razdelitvijo notebooka tukaj nastavite javni HTTPS naslov checkerja.
# Primer: https://checker.moja-domena.si/preveri
CHECKER_URL = "https://fe-lab-checker.onrender.com/preveri"


def _za_json(vrednost):
    """Pretvori numpy/pandas vrednosti v obliko, ki jo je mogoce poslati kot JSON."""
    if vrednost is Ellipsis:
        return None
    if isinstance(vrednost, dict):
        return {str(k): _za_json(v) for k, v in vrednost.items()}
    if isinstance(vrednost, (list, tuple, set)):
        return [_za_json(v) for v in vrednost]
    if hasattr(vrednost, "tolist"):
        try:
            return _za_json(vrednost.tolist())
        except Exception:
            pass
    if hasattr(vrednost, "to_dict"):
        try:
            return _za_json(vrednost.to_dict())
        except Exception:
            pass
    if hasattr(vrednost, "item"):
        try:
            return _za_json(vrednost.item())
        except Exception:
            pass
    if hasattr(vrednost, "isoformat"):
        try:
            return vrednost.isoformat()
        except Exception:
            pass
    return vrednost


def _najdi_placeholderje(vrednost, pot="", _seen=None):
    """Poisce (...) tudi znotraj seznamov, slovarjev in numpy/pandas objektov."""
    if vrednost is Ellipsis:
        return [pot or "<vrednost>"]

    if _seen is None:
        _seen = set()

    # Preprecimo tezave pri morebitnih samoreferencnih strukturah.
    try:
        oid = id(vrednost)
        if oid in _seen:
            return []
        _seen.add(oid)
    except Exception:
        pass

    najdeni = []

    if isinstance(vrednost, dict):
        for k, v in vrednost.items():
            najdeni.extend(_najdi_placeholderje(v, f"{pot}[{k!r}]", _seen))
        return najdeni

    if isinstance(vrednost, (list, tuple)):
        for i, v in enumerate(vrednost):
            najdeni.extend(_najdi_placeholderje(v, f"{pot}[{i}]", _seen))
        return najdeni

    if isinstance(vrednost, set):
        for i, v in enumerate(vrednost):
            najdeni.extend(_najdi_placeholderje(v, f"{pot}[{i}]", _seen))
        return najdeni

    # numpy/pandas objekti
    if hasattr(vrednost, "tolist"):
        try:
            pretvorjeno = vrednost.tolist()
            if pretvorjeno is not vrednost:
                return _najdi_placeholderje(pretvorjeno, pot, _seen)
        except Exception:
            pass

    if hasattr(vrednost, "to_dict"):
        try:
            pretvorjeno = vrednost.to_dict()
            if pretvorjeno is not vrednost:
                return _najdi_placeholderje(pretvorjeno, pot, _seen)
        except Exception:
            pass

    return najdeni


def preveri():
    # Spremenljivke, ki morajo biti definirane v predhodnih nalogah.
    zahtevane = [
        "ime", "vpisna_st", "vpisna_st_2", "faktor_vpisna", "kot_rad",
        "tg_kota", "lst_vpisna", "pov_vpisna", "stolpec_sestevek", "min_amp_i"
    ]

    manjkajo = [ime_sprem for ime_sprem in zahtevane if ime_sprem not in globals()]
    if manjkajo:
        print("❌ Preverjanje ni mogoce. Najprej definirajte vse zahtevane spremenljivke:")
        for ime_sprem in manjkajo:
            print(f"   - {ime_sprem}")
        return

    # Preverimo (...) tudi znotraj seznamov/arrayev/slovarjev.
    nedokoncane = []
    for ime_sprem in zahtevane:
        poti = _najdi_placeholderje(globals()[ime_sprem], ime_sprem)
        nedokoncane.extend(poti)

    # Uciteljska resitev uporablja tudi lst_vpisna_lihe; ce obstaja, preverimo se to.
    if "lst_vpisna_lihe" in globals():
        nedokoncane.extend(_najdi_placeholderje(globals()["lst_vpisna_lihe"], "lst_vpisna_lihe"))

    if nedokoncane:
        print("❌ Vaja se ni pripravljena za preverjanje.")
        print("Placeholder (...) je ostal na naslednjih mestih:")
        for pot in nedokoncane:
            print(f"   - {pot}")
        print("\nDopolnite te vrednosti in nato ponovno pozenite celico PREVERJANJE.")
        return

    if "TUKAJ-VPISI-NASLOV-CHECKERJA" in CHECKER_URL:
        print("❌ Checker se ni nastavljen. Obvestite izvajalca laboratorijske vaje.")
        return

    vrednosti = {ime_sprem: _za_json(globals()[ime_sprem]) for ime_sprem in zahtevane}

    # Uciteljska resitev uporablja tudi lst_vpisna_lihe; ce obstaja, jo posljemo dodatno.
    if "lst_vpisna_lihe" in globals():
        vrednosti["lst_vpisna_lihe"] = _za_json(globals()["lst_vpisna_lihe"])

    pricakovane_datoteke = [
        "primer_shranjevanja.csv", "graf_KD.png", "graf_histogram.png",
        "graf_amp_u.png", "graf_KD_2.png", "graf_amp_fi_i.png", "resitve_podatki.csv"
    ]
    datoteke = {
        ime_dat: os.path.exists(os.path.join("Rezultati", ime_dat))
        for ime_dat in pricakovane_datoteke
    }

    # Grafi so samo informativni. Posljemo vseh pet PNG-jev, ce obstajajo.
    # Noben rezultat slikovnega preverjanja ne vpliva na kodo za eUcilnico.
    grafi_za_preverjanje = [
        "graf_histogram.png",
        "graf_KD.png",
        "graf_amp_u.png",
        "graf_KD_2.png",
        "graf_amp_fi_i.png",
    ]
    slike = {}
    for ime_grafa in grafi_za_preverjanje:
        pot_grafa = os.path.join("Rezultati", ime_grafa)
        if not os.path.exists(pot_grafa):
            continue

        velikost = os.path.getsize(pot_grafa)
        if velikost <= 4_000_000:
            with open(pot_grafa, "rb") as f:
                slike[ime_grafa] = base64.b64encode(f.read()).decode("ascii")
        else:
            print(f"⚠️ {ime_grafa} je vecji od 4 MB in ga slikovni checker ne bo poslal.")

    try:
        vpisna_za_posiljanje = int(vpisna_st)
    except (TypeError, ValueError):
        print("❌ Spremenljivka vpisna_st mora vsebovati veljavno vpisno stevilko.")
        print(f"Trenutna vrednost je: {vpisna_st!r}")
        return

    paket = {
        "vaja": "UV0",
        "vpisna_st": vpisna_za_posiljanje,
        "values": vrednosti,
        "files": datoteke,
        "images": slike,
    }

    try:
        podatki = json.dumps(paket, ensure_ascii=False).encode("utf-8")
    except TypeError as e:
        print("❌ Ene od rezultatov ni mogoce poslati checkerju.")
        print(f"Podrobnost: {e}")
        print("Preverite, da zahtevane spremenljivke vsebujejo stevila, besedilo, sezname ali slovarje.")
        return

    zahteva = urllib.request.Request(
        CHECKER_URL,
        data=podatki,
        headers={"Content-Type": "application/json"},
        method="POST",
    )

    print("Preverjam rezultate ...\n")

    try:
        with urllib.request.urlopen(zahteva, timeout=150) as odgovor:
            rezultat = json.loads(odgovor.read().decode("utf-8"))
    except urllib.error.HTTPError as e:
        print(f"❌ Checker je vrnil napako HTTP {e.code}.")
        return
    except urllib.error.URLError:
        print("❌ Checker trenutno ni dosegljiv. Preverite internetno povezavo in poskusite znova.")
        return
    except Exception as e:
        print(f"❌ Preverjanje ni uspelo: {e}")
        return

    if rezultat.get("server_error"):
        print("⚠️ Napaka na checker strezniku:")
        print(rezultat["server_error"])
        return

    print("=" * 52)
    print("                 REZULTAT PREVERJANJA")
    print("=" * 52)

    for kontrola in rezultat.get("checks", []):
        znak = "✅" if kontrola.get("ok") else "❌"
        print(f"{znak} {kontrola.get('name')}: {kontrola.get('message', '')}")

    print("-" * 52)
    print(f"Pravilno: {rezultat.get('passed', 0)} / {rezultat.get('total', 0)}")

    # Grafi/datoteke so samo informativni in NE vplivajo na kodo.
    advisory = rezultat.get("advisory", [])
    if advisory:
        print("\n" + "=" * 52)
        print("        GRAFI IN DATOTEKE - INFORMATIVNO")
        print("        (NE vplivajo na izdajo kode)")
        print("=" * 52)
        for ocena in advisory:
            status = ocena.get("status")
            if status == "ok":
                znak = "✅"
            elif status == "likely_wrong":
                znak = "❌"
            elif status == "warning":
                znak = "⚠️"
            else:
                znak = "❔"
            print(f"{znak} {ocena.get('name')}: {ocena.get('message', '')}")
        print("-" * 52)

    if rezultat.get("all_ok"):
        print("\n🎉 Vaja je uspesno opravljena!")
        print("\nKoda za eUcilnico:")
        print("=" * 52)
        print(rezultat.get("code", ""))
        print("=" * 52)
        print("Kodo kopirajte v eUcilnico.")
    else:
        print("\nPopravite oznacene naloge in ponovno pozenite to celico.")


# Samodejno preverjanje ob zagonu celice.
preveri()
```
